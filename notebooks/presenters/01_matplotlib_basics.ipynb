# Matplotlib Basics with Citi Bike

Today we learn the Matplotlib mental model and build several chart types for our Citi Bike questions.


**Learning objectives**
- Understand Figure vs Axes
- Build line, bar, histogram, and scatter plots
- Pick the right chart for the question
- Save figures to disk

## Setup


```python
from pathlib import Path
import sys

BASE_DIR = Path.cwd()
if (BASE_DIR / "viz-workshop").exists():
    BASE_DIR = BASE_DIR / "viz-workshop"
elif (BASE_DIR / "notebooks").exists():
    BASE_DIR = BASE_DIR.parent

sys.path.append(str(BASE_DIR / "../../src"))

from data_prep import load_or_create_dataset, get_paths, print_environment_hint

print_environment_hint()
paths = get_paths(BASE_DIR)
```

```python
# Load or build the processed dataset (cached after first run)
df = load_or_create_dataset(base_dir=BASE_DIR)
print(df.shape)
df.head()
```

## Episode 1: Figure vs Axes

Think of **Figure** as the whole canvas and **Axes** as one plot area.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 3))
ax.plot(df["hour"].value_counts().sort_index().index,
        df["hour"].value_counts().sort_index().values)
ax.set_title("Trips by Hour (quick view)")
ax.set_xlabel("Hour of day")
ax.set_ylabel("Trips")
plt.tight_layout()
plt.show()
```

## Episode 2: Line plot (daily trips)

Line plots are great for time series.


```python
trips_by_day = df.groupby("date").size().reset_index(name="trips")

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(trips_by_day["date"], trips_by_day["trips"], color="#2ca02c")
ax.set_title("Daily Citi Bike Trips")
ax.set_xlabel("Date")
ax.set_ylabel("Trips")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Optional mini-extension:** try a 7-day rolling average.


```python
trips_by_day["rolling_7d"] = trips_by_day["trips"].rolling(7).mean()

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(trips_by_day["date"], trips_by_day["trips"], alpha=0.4, label="daily")
ax.plot(trips_by_day["date"], trips_by_day["rolling_7d"], linewidth=2, label="7-day avg")
ax.legend()
ax.set_title("Daily Trips with Rolling Average")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Episode 3: Bar charts

Bar charts are good for comparing categories.


```python
member_counts = df["member_casual"].value_counts()

fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(member_counts.index, member_counts.values, color=["#1f77b4", "#ff7f0e"])
ax.set_title("Member vs Casual Trips")
ax.set_xlabel("Rider type")
ax.set_ylabel("Trips")
plt.tight_layout()
plt.show()
```

```python
# Top 10 start stations
station_counts = df["start_station_name"].value_counts().head(10).sort_values()

fig, ax = plt.subplots(figsize=(6, 4))
ax.barh(station_counts.index, station_counts.values, color="#9467bd")
ax.set_title("Top 10 Start Stations")
ax.set_xlabel("Trips")
plt.tight_layout()
plt.show()
```

## Episode 4: Histograms

Histograms show the distribution of a numeric variable like trip duration.


```python
fig, ax = plt.subplots(figsize=(6, 3))
ax.hist(df["duration_min"], bins=40, color="#8c564b", alpha=0.8)
ax.set_title("Trip Duration (minutes)")
ax.set_xlabel("Duration (min)")
ax.set_ylabel("Trips")
plt.tight_layout()
plt.show()
```

## Episode 5: Scatter plots

Scatter plots help show relationships, but can suffer from overplotting.


```python
sample = df.sample(3000, random_state=42)

fig, ax = plt.subplots(figsize=(6, 3))
ax.scatter(sample["hour"], sample["duration_min"], alpha=0.3)
ax.set_title("Hour vs Duration (sampled)")
ax.set_xlabel("Hour of day")
ax.set_ylabel("Duration (min)")
plt.tight_layout()
plt.show()
```

## Episode 6: Saving plots

Use `savefig` to export. Here we save into `outputs/`.


```python
from pathlib import Path

outputs_dir = BASE_DIR / "outputs"
outputs_dir.mkdir(exist_ok=True)

fig, ax = plt.subplots(figsize=(6, 3))
ax.hist(df["duration_min"], bins=50, color="#2ca02c")
ax.set_title("Trip Duration (saved)")

fig.savefig(outputs_dir / "duration_hist.png", dpi=150, bbox_inches="tight")
plt.show()
```

**Solution (optional): weekday usage bar chart**

```python
weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
weekday_counts = df.groupby("day_name").size().reindex(weekday_order)

fig, ax = plt.subplots(figsize=(6, 3))
ax.bar(weekday_counts.index, weekday_counts.values, color="#1f77b4")
ax.set_title("Trips by Day of Week")
ax.set_ylabel("Trips")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Common pitfalls
- Overplotting: use `alpha` and sampling for dense scatter plots.
- Axis confusion: always label axes, especially for time series.
